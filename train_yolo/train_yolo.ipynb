# Merge Datasets

```python
import os
import shutil
import yaml

# 1. Define the datasets to merge (folder names)
DATASETS = ["Bowling-1", "car-detection-5", "My-First-Project-4", "Toy-car-detection-3"]

MERGED_DIR = "Merged_Dataset"


def get_yaml_path(dataset_path):
    """Find the yaml file in the dataset directory."""
    for file in os.listdir(dataset_path):
        if file.endswith(".yaml"):
            return os.path.join(dataset_path, file)
    raise FileNotFoundError(f"No YAML file found in {dataset_path}")


def main():
    # 2. Extract all unique classes across all datasets to build the global map
    global_classes = []

    for ds in DATASETS:
        yaml_path = get_yaml_path(ds)
        with open(yaml_path, "r") as f:
            data = yaml.safe_load(f)
            for cls_name in data["names"]:
                if cls_name not in global_classes:
                    global_classes.append(cls_name)

    print(f"Global classes mapped: {global_classes}")

    # 3. Create the Merged Dataset structure
    splits = ["train", "valid", "test"]
    for split in splits:
        os.makedirs(os.path.join(MERGED_DIR, split, "images"), exist_ok=True)
        os.makedirs(os.path.join(MERGED_DIR, split, "labels"), exist_ok=True)

    # 4. Process each dataset
    for ds in DATASETS:
        print(f"Processing dataset: {ds}...")
        yaml_path = get_yaml_path(ds)

        with open(yaml_path, "r") as f:
            data = yaml.safe_load(f)
            local_classes = data["names"]  # list of class names where index == local ID

        # Create a mapping dictionary for this specific dataset: local_id -> global_id
        local_to_global_map = {}
        for local_id, cls_name in enumerate(local_classes):
            local_to_global_map[local_id] = global_classes.index(cls_name)

        # Process each split (train, valid, test)
        for split in splits:
            images_dir = os.path.join(ds, split, "images")
            labels_dir = os.path.join(ds, split, "labels")

            if not os.path.exists(images_dir):
                continue  # Skip if this split doesn't exist in the dataset

            for img_name in os.listdir(images_dir):
                if not img_name.lower().endswith((".png", ".jpg", ".jpeg")):
                    continue

                # Prefix filename to avoid collisions between datasets
                new_img_name = f"{ds}_{img_name}"
                src_img_path = os.path.join(images_dir, img_name)
                dst_img_path = os.path.join(MERGED_DIR, split, "images", new_img_name)

                # Copy image
                shutil.copy2(src_img_path, dst_img_path)

                # Process corresponding label file
                label_name = img_name.rsplit(".", 1)[0] + ".txt"
                new_label_name = f"{ds}_{label_name}"
                src_label_path = os.path.join(labels_dir, label_name)
                dst_label_path = os.path.join(
                    MERGED_DIR, split, "labels", new_label_name
                )

                if os.path.exists(src_label_path):
                    with (
                        open(src_label_path, "r") as f_in,
                        open(dst_label_path, "w") as f_out,
                    ):
                        for line in f_in:
                            parts = line.strip().split()
                            if len(parts) == 0:
                                continue

                            local_id = int(parts[0])

                            # Replace local ID with global ID
                            if local_id in local_to_global_map:
                                global_id = local_to_global_map[local_id]
                                parts[0] = str(global_id)
                                f_out.write(" ".join(parts) + "\n")

    # 5. Generate the new merged data.yaml
    merged_yaml_content = {
        "path": os.path.abspath(MERGED_DIR),
        "train": "train/images",
        "val": "valid/images",
        "test": "test/images",
        "nc": len(global_classes),
        "names": global_classes,
    }

    with open(os.path.join(MERGED_DIR, "data.yaml"), "w") as f:
        yaml.dump(merged_yaml_content, f, sort_keys=False)

    print(f"\nSuccessfully merged all datasets into '{MERGED_DIR}'.")
    print(
        f"You can now train using: model.train(data='{MERGED_DIR}/data.yaml', epochs=100, imgsz=640, device='cuda')"
    )


if __name__ == "__main__":
    main()
```

# Finetune on Our Data

```python

# --- CONFIGURATION ---
# Path to the root of your newly downloaded Roboflow dataset
NEW_DATASET_PATH = "/home/youssef_mohammad/projects/Aba-ViTrack/train_yolo/my_dataset"  # CHANGE THIS to your folder name

# The mapping from the NEW dataset's IDs to the MERGED dataset's IDs
# Format: { Local_ID : Global_ID }
ID_MAPPING = {
    0: 0,  # bowling-ball -> bowling-ball (No change, but safe to include)
    1: 1,  # bowling-pins -> bowling-pins (No change)
    2: 3,  # car -> car (Crucial change!)
}

# The global class list that matches your pre-trained model exactly
GLOBAL_CLASSES = ["bowling-ball", "bowling-pins", "sweep board", "car"]
# ---------------------


def align_dataset():
    splits = ["train", "valid", "test"]

    # 1. Update the label files
    for split in splits:
        labels_dir = os.path.join(NEW_DATASET_PATH, split, "labels")

        # Some splits (like test) might not exist, skip if so
        if not os.path.exists(labels_dir):
            continue

        print(f"Updating labels in {labels_dir}...")

        for filename in os.listdir(labels_dir):
            if not filename.endswith(".txt"):
                continue

            filepath = os.path.join(labels_dir, filename)

            with open(filepath, "r") as file:
                lines = file.readlines()

            new_lines = []
            for line in lines:
                parts = line.strip().split()
                if not parts:
                    continue

                local_id = int(parts[0])

                if local_id in ID_MAPPING:
                    global_id = ID_MAPPING[local_id]
                    parts[0] = str(global_id)

                new_lines.append(" ".join(parts) + "\n")

            with open(filepath, "w") as file:
                file.writelines(new_lines)

    # 2. Update the data.yaml file to match the 4-class structure
    yaml_path = os.path.join(NEW_DATASET_PATH, "my_dataset.yaml")
    if os.path.exists(yaml_path):
        print(f"Updating {yaml_path}...")
        with open(yaml_path, "r") as file:
            yaml_data = yaml.safe_load(file)

        # Overwrite the class configuration
        yaml_data["nc"] = len(GLOBAL_CLASSES)
        yaml_data["names"] = GLOBAL_CLASSES

        with open(yaml_path, "w") as file:
            yaml.dump(yaml_data, file, sort_keys=False)

    print("\n✅ Dataset alignment complete!")
    print(
        "All 'car' labels are now '3', and your data.yaml perfectly matches your trained model."
    )


if __name__ == "__main__":
    align_dataset()
```
